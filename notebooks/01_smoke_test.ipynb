## Checking Python version and pacakges

```python
import sys, importlib

print("Python:", sys.version.splitlines()[0])

def check(name):
    try:
        m = importlib.import_module(name)
        print(f"{name}: {getattr(m, '__version__', 'n/a')}")
    except Exception as e:
        print(f"{name}: import failed -> {e}")
        
        
check("TTS")
check("onnxruntime")

try:
    import torch
    print("torch:", torch.__version__, "cuda_available:", torch.cuda.is_available())
except Exception as e:
    print("torch: import failed ->", e)
```

## Checking TTS

```python
try:
    import traceback
    from TTS.api import TTS
    
    MODEL = "tts_models/pt/cv/vits"
    
    try:
        print("Loading model:", MODEL)
        tts = TTS(model_name=MODEL)
        wav = tts.tts("Olá mundo")
        import soundfile as sf
        sr = getattr(getattr(tts, "synthesizer", None), "output_sample_rate", 22050)
        sf.write("../data/outputs/sample_ola.wav", wav, sr)
        print("Wrote sample_ola.wav")
    except Exception:
        print("Model load / synth failed, full traceback below:")
        traceback.print_exc()
except Exception as e:
    print("TTS test error:", e)
```
